## Step 3: Get to know the free tier

Nothing to set up here, but these limits explain how the free route behaves later in the week. They come from Render's page on [free instances](https://render.com/docs/free), where you can check the current details.

- **Free web services spin down when idle.** After 15 minutes without a request, Render stops the service. The next request starts it again, which takes about a minute, and Render shows a loading page meanwhile. So expect the first request of a session to be slow.
- **750 free instance hours a month, shared across your workspace.** Our services spin down when nobody is using them, so a week of labs uses far fewer hours than that. If you ever use them all, Render suspends your free services until the next month starts.
- **Nothing is kept on the service's own disk.** Free web services have no persistent disk, so anything a service writes to its disk is lost when it restarts. That's fine for us, because UnderCover keeps its data in Supabase.
- **No shell access and a single instance.** You can't open a shell on a free service, and it can't scale beyond one instance. Its logs are in the Render dashboard.
- **Use Supabase for the database.** Render offers free Postgres databases too, but they expire 30 days after creation, so we stay with the Supabase project from Week 2.
- **Payment.** If you haven't added a payment method and a free service goes over the free bandwidth or build minutes, Render suspends it or stops its builds rather than charging you. If you add one, usage beyond the free amounts is billed. The amounts are on Render's [pricing page](https://render.com/pricing).
- **Not for production.** Render says free instances shouldn't be used for production applications. They're right for a course.

## Solution - assignment challenge

### Discovery note: first-pass review of supplier invoice exceptions

*Illustrative business, chosen for this assignment: Meridian Freight, a 400-person logistics company. The figures below are the baselines I would confirm with a two-week measurement before committing to the targets; nothing here has been measured on a live system.*

**Who has the problem.** The accounts-payable team: three clerks who work the exception queue, the AP manager who handles escalations and approves large payments, and the controller who closes the books. Suppliers and internal requesters also feel it: they wait for answers and for payment. The AP manager owns the problem; the controller owns the cost.

**What the problem is today.** Around 45 supplier invoices a week arrive as exceptions: no matching purchase order, a price or quantity mismatch, a duplicate of an invoice already paid, or a missing budget approval. A clerk opens the ERP screen, the PDF in the inbox, the email thread, the vendor master and the contract rate card, copies references between them, and asks the requester for context by email. The reason for the final decision is usually only in the clerk's reply, not in a record. Similar cases are decided differently by different clerks, and month-end close waits for the backlog to clear.

**What it costs today (baseline to confirm).**

- Handling time: about 45 exceptions × 22 minutes = 16.5 hours a week, roughly 0.4 of a full-time clerk, at a loaded $60/hour, about $43,000 a year.
- Late-payment fees on invoices that sit in the queue: about $8,000 a year.
- Early-payment discounts missed because approval takes days: about $14,000 a year.
- Duplicate or incorrect payments found at audit: about $12,000 a quarter.
- Close delayed by roughly 1.5 working days each month, which is the controller's main complaint.

**What an agent would do.** For each exception it reads the invoice, the purchase order, the receiving record, the customer's (supplier's) earlier exceptions, and the sections of the expense and procurement policy it needs. It returns `approve`, `decline` or `refer`, the approved amount with its calculation, and a two- or three-sentence explanation naming the policy section. It releases approved amounts up to $5,000 itself; above that it pauses for the AP manager's sign-off, who either approves or sends it back with a note. It records every lookup, tool call and refusal so the controller can audit the decision, and it keeps a per-supplier memory of earlier decisions and manager notes so a later exception has context.

**Success criteria, with a baseline and a target for each.**

| Measure | How I would measure it | Baseline | Target at 90 days |
| --- | --- | --- | --- |
| Straight-through rate | Share of exceptions decided and paid without a human edit, over a rolling 500 cases | 0% (a clerk touches every case) | at least 60% |
| Median time from arrival to decision | ERP timestamps, eligible exceptions only | 3.5 days | 4 hours or less |
| Decision accuracy | Agreement with a 200-invoice set labelled by two senior clerks; the agent never sees this set | 88% clerk-to-clerk agreement on hard cases | at least 95% agreement |
| Amount accuracy | Approved amount within $1 of the manager's amount on the labelled set | 91% | at least 98% |
| Cost per exception | Loaded labour plus model and infrastructure cost, from the run record | about $22 | $7 or less |
| Duplicate or incorrect payment loss | Findings in the quarterly audit | about $12,000/quarter | under $3,000/quarter |
| Late-payment fees | Penalty lines in the ERP | about $8,000/year | under $2,000/year |
| Audit completeness | Share of decisions carrying reason, section, maker, time and policy version | about 70% | 100% of agent decisions |
| Manager override rate | Share of auto-decisions later reversed | not applicable, no automation | under 10% |
| Sign-off SLA above $5,000 | Median time from hold to the manager's answer | 1.5 days in the current routing | 4 business hours or less |

I would take the baseline from a two-week log pull and the labelled sample before writing a line of code, agree the targets with the controller, then review the measures weekly during a four-week pilot. A referral that the policy does not support is a failure, not a success: the straight-through target must not be met by referring everything.

**Scope and guardrails.** The agent only reads the ERP and policy files; it never edits the vendor master or sends supplier emails. The approval limit lives in a settings record, not in the prompt, and the payment endpoint refuses an over-limit amount before it reads the invoice. Senior approval is the only way an over-limit payment is released. The labelled set is held out from the agent, and evaluations use no memory. If the agent is wrong twice, the case goes to a human and stays there.

### Stretch: architecture decision record (summary)

- **Decision.** Build a thin agent over the ERP's existing APIs rather than buying a new AP automation module, and keep the human sign-off in our own app.
- **Options considered.** (A) Buy the ERP vendor's exception-matching add-on: quickest to run, but its rules cannot read an email thread or a changing contract rate card, and its decision log does not meet our audit format. (B) Build an agent that reads the same systems through our API and a read-only policy MCP server, with checkpoints in Postgres: more work, but we own the decision record, the limits and the evaluation set. (C) Keep the manual queue: no build risk, but none of the success targets move.
- **Choice and where it runs.** Option B. During the course it deploys on the free route: a FastAPI web service and a dashboard on Render, the agent as a second free web service with the context MCP server beside it, and Supabase for records and checkpoints. For a client I would use the same code in a managed container behind an API gateway with a policy engine, rather than a free instance. The agent needs to reach the ERP read API, the document store, the email archive, the policy files and the approval service; the policy files have their own owner and are read-only to the agent.
- **Consequences.** We keep control of the rules and the audit trail, and can roll out in shadow/LOG_ONLY mode before releasing payments. We take on the work of testing, monitoring and evaluating an agent, and must budget for model calls and for the latency of a free service that sleeps when idle.
